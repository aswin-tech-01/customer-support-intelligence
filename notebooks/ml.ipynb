```python
df["Ticket Subject"] = df["Ticket Subject"].str.replace(r"\s+", " ", regex=True).str.strip()

df["Ticket Description"] = df["Ticket Description"].str.replace(r"\s+", " ", regex=True).str.strip()

print(df["Ticket Subject"].head(5))
print("\n")                                      #this step is intended to remove unc
print(df["Ticket Description"].head(5))
```

```python
df["Ticket Text"] = (
    df["Ticket Subject"].fillna("").astype(str)
    + " "
    + df["Ticket Description"].fillna("").astype(str)
)

print(df["Ticket Text"].head(5).to_string())
```

```python
df["text_length"] = df["Ticket Text"].str.len()
df["word_count"] = df["Ticket Text"].str.split().str.len()

print("Average text length:", df["text_length"].mean())
print("Average word count:", df["word_count"].mean())

print("\nText statistics:")
print(df[["text_length", "word_count"]].describe())
```

```python
from sklearn.model_selection import train_test_split

X = df["Ticket Text"]
y = df["Ticket Type"]
                                            #here we are dividin the data into training and testing data 
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y             #ensures the same approximate proportion of each ticket type exists in both training and testing data.
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(                         #we converted the customer complaints into numerical values that can be used for ml 
    max_features=5000,
    stop_words="english"
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)                                           #This is our first actual predictive component of the project.

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test_tfidf)
                                             #here we uses 20% of data 
print("Predictions completed!")
print("\nFirst 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
def calculate_priority(text):
    text = str(text).lower()

    critical_keywords = [
        "urgent",
        "immediately",
        "critical",
        "emergency",
        "completely stopped",
        "cannot access",
        "unable to access"
    ]

    high_keywords = [
        "not working",
        "stopped working",
        "failed",
        "failure",
        "severe",
        "serious",
        "cannot use",
        "unable to use"
    ]

    medium_keywords = [
        "problem",
        "issue",
        "error",
        "slow",
        "delay",
        "difficult"
    ]

    if any(keyword in text for keyword in critical_keywords):
        return "Critical"

    elif any(keyword in text for keyword in high_keywords):
        return "High"

    elif any(keyword in text for keyword in medium_keywords):
        return "Medium"

    else:
        return "Low"


print("Priority scoring function created successfully!")
```

```python
from sentence_transformers import SentenceTransformer

semantic_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Semantic model loaded successfully!")
```

```python
severity_examples = {
    "Critical": [
        "The entire service is down and I cannot access anything.",
        "I am completely locked out of my account and cannot access my important data.",
        "There is a serious security problem and my account may be compromised.",
        "I have lost critical data and urgently need assistance.",
        "The system has completely stopped working and is causing major disruption."
    ],
    
    "High": [
        "An important feature has stopped working and I cannot use it.",
        "I am unable to access an important part of the product.",
        "The product keeps failing and I cannot use it properly.",
        "A major functionality is not working despite trying several times.",
        "This issue is seriously affecting my ability to use the product."
    ],
    
    "Medium": [
        "I am experiencing a problem with the product.",
        "The product is not working properly sometimes.",
        "I am facing an issue but I can still use the product.",
        "The application is slow and occasionally gives errors.",
        "There is a recurring problem that is affecting my experience."
    ],
    
    "Low": [
        "I would like some information about the product.",
        "Can you tell me more about this product?",
        "I need help understanding how this product works.",
        "I would like a product recommendation.",
        "I have a general question about the product."
    ]
}

print("Severity reference examples created successfully!")

for priority, examples in severity_examples.items():
    print(f"{priority}: {len(examples)} examples")
```

```python
severity_embeddings = {}

for priority, examples in severity_examples.items():
    severity_embeddings[priority] = semantic_model.encode(
        examples,
        convert_to_tensor=True
    )

print("Severity embeddings created successfully!")

for priority, embeddings in severity_embeddings.items():
    print(priority, "→", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def predict_priority(text):
    # Convert the customer complaint into an embedding
    complaint_embedding = semantic_model.encode(
        [text],
        convert_to_tensor=True
    )

    scores = {}

    # Compare complaint with every priority level
    for priority, embeddings in severity_embeddings.items():
        similarity = cosine_similarity(
            complaint_embedding.cpu().numpy(),
            embeddings.cpu().numpy()
        )

        scores[priority] = similarity.mean()

    # Select the priority with the highest similarity
    predicted_priority = max(scores, key=scores.get)

    return predicted_priority, scores
```

```python
test_complaints = [
    "My entire account is locked and I cannot access any of my important data.",
    "My device has stopped working and I cannot use it.",
    "The application is slow and sometimes gives errors.",
    "I would like some information about this product."
]

for complaint in test_complaints:
    priority, scores = predict_priority(complaint)

    print("\nComplaint:", complaint)
    print("Predicted Priority:", priority)
    print("Scores:", {k: round(v, 3) for k, v in scores.items()})
```

```python
# Create semantic embeddings for the training tickets

from collections import defaultdict
import numpy as np

# Encode all training ticket texts
X_train_embeddings = semantic_model.encode(
    X_train_root.tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Training embeddings shape:", X_train_embeddings.shape)
```

```python
# Create one semantic prototype for each issue category

issue_prototypes = {}

for issue in y_train_root.unique():
    issue_embeddings = X_train_embeddings[y_train_root.values == issue]
    
    issue_prototypes[issue] = issue_embeddings.mean(axis=0)

print("Number of issue prototypes:", len(issue_prototypes))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def predict_issue_semantic(text):
    # Convert the new complaint into an embedding
    complaint_embedding = semantic_model.encode(
        [text],
        convert_to_numpy=True
    )

    scores = {}

    # Compare complaint with every issue prototype
    for issue, prototype in issue_prototypes.items():
        similarity = cosine_similarity(
            complaint_embedding,
            prototype.reshape(1, -1)
        )[0][0]

        scores[issue] = similarity

    # Select the issue with the highest similarity
    predicted_issue = max(scores, key=scores.get)

    return predicted_issue, scores
```

```python
test_complaints = [
    "My order has not arrived yet.",
    "I was charged twice for the same order.",
    "I want to cancel my order.",
    "I cannot login to my account.",
    "My phone battery drains very quickly."
]

for complaint in test_complaints:
    predicted_issue, scores = predict_issue_semantic(complaint)

    print("\nComplaint:", complaint)
    print("Predicted Issue:", predicted_issue)
```
